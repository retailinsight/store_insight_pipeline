```python
import sys, os

_root = os.path.abspath(os.path.join('..', '..', '..'))
if _root not in sys.path:
    sys.path.insert(0, _root)

import store_insight.alerts  # registers alerts_pipeline alias automatically

from store_insight.alerts.pipeline.session import load_env, create_session

load_env()
session = create_session()
print('Snowflake session created.')
```

```python
_SCHEMA = "INBOUND.SOURCE"

WMD_TABLES = [
    "SOURCE_WMD_ITEM_HIERARCHY",
    "SOURCE_WMD_MARKDOWN_LOG_CONVERTED",
    "SOURCE_WMD_MARKDOWN_SALES",
    "SOURCE_WMD_PPG_CLUSTER_MAPPING",
    "SOURCE_WMD_SALES",
    "SOURCE_WMD_STORE_CLUSTER_MAPPING",
    "SOURCE_WMD_STORE_HIERARCHY",
    "SOURCE_WMD_STORE_TRADING_HOUR",
    "SOURCE_WMD_STORE_TRADING_HOUR_HOLIDAY",
    "SOURCE_WMD_WASTAGE",
]

print(f"{len(WMD_TABLES)} tables to explore in {_SCHEMA}")
```

```python
def explore_table(table_name):
    """Print schema, row count, and sample rows for a WMD table."""
    fqn = f"{_SCHEMA}.{table_name}"
    df = session.table(fqn)

    # Schema
    print(f"\n{'=' * 80}")
    print(f"  {table_name}")
    print(f"{'=' * 80}")
    print(f"\nColumns ({len(df.schema.fields)}):")
    for f in df.schema.fields:
        print(f"  {f.name:<45} {str(f.datatype):<20} nullable={f.nullable}")

    # Row count
    count = df.count()
    print(f"\nRow count: {count:,}")

    # Sample
    print(f"\nSample (5 rows):")
    df.show(5)

    return df
```
